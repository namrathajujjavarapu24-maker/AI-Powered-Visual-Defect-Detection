```python
import gradio as gr
from deep_translator import GoogleTranslator


# ============================================================
# LANGUAGES
# ============================================================

languages = {
    "English": "en",
    "Hindi": "hi",
    "Telugu": "te",
    "Tamil": "ta",
    "Kannada": "kn",
    "Malayalam": "ml",
    "French": "fr",
    "German": "de",
    "Spanish": "es",
    "Japanese": "ja",
    "Chinese": "zh-CN",
    "Korean": "ko"
}


# ============================================================
# TRANSLATION FUNCTION
# ============================================================
import requests

def translate_text(text, source_language, target_language):

    if not text.strip():
        return "⚠️ Please enter some text first."

    if source_language == target_language:
        return text

    try:
        source_code = languages[source_language]
        target_code = languages[target_language]

        url = "https://api.mymemory.translated.net/get"

        params = {
            "q": text,
            "langpair": f"{source_code}|{target_code}"
        }

        response = requests.get(
            url,
            params=params,
            timeout=15
        )

        if response.status_code != 200:
            return "❌ Translation service is temporarily unavailable."

        data = response.json()

        translated_text = data["responseData"]["translatedText"]

        if not translated_text:
            return "❌ No translation was returned."

        return translated_text

    except requests.exceptions.Timeout:
        return "⏳ Translation request timed out. Please try again."

    except Exception as e:
        return f"❌ Translation error: {str(e)}"
# ============================================================
# SWAP LANGUAGES
# ============================================================

def swap_languages(source, target):

    return target, source


# ============================================================
# CLEAR
# ============================================================

def clear_all():

    return "", "", "English", "Hindi"


# ============================================================
# CHARACTER COUNTER
# ============================================================

def character_count(text):

    return f"{len(text)} characters"


# ============================================================
# CUSTOM CSS
# ============================================================

custom_css = """

/* =========================================================
   GLOBAL
========================================================= */

* {
    box-sizing: border-box;
}

body {
    margin: 0;
    background: #070b1a;
}

.gradio-container {
    max-width: 1200px !important;
    margin: auto !important;
    padding: 0 !important;
    font-family: Inter, "Segoe UI", Arial, sans-serif !important;
    background:
        radial-gradient(circle at 10% 10%, rgba(99,102,241,0.18), transparent 30%),
        radial-gradient(circle at 90% 20%, rgba(168,85,247,0.15), transparent 30%),
        radial-gradient(circle at 50% 100%, rgba(59,130,246,0.12), transparent 35%),
        #070b1a !important;
}


/* =========================================================
   ANIMATED BACKGROUND
========================================================= */

.gradio-container::before {
    content: "";
    position: fixed;
    width: 450px;
    height: 450px;
    border-radius: 50%;
    background: rgba(99,102,241,0.10);
    filter: blur(80px);
    top: 5%;
    left: -150px;
    animation: floatOne 8s ease-in-out infinite alternate;
    pointer-events: none;
}

.gradio-container::after {
    content: "";
    position: fixed;
    width: 400px;
    height: 400px;
    border-radius: 50%;
    background: rgba(168,85,247,0.10);
    filter: blur(80px);
    right: -150px;
    bottom: 5%;
    animation: floatTwo 10s ease-in-out infinite alternate;
    pointer-events: none;
}

@keyframes floatOne {

    from {
        transform: translateY(0px);
    }

    to {
        transform: translateY(80px);
    }

}

@keyframes floatTwo {

    from {
        transform: translateY(0px);
    }

    to {
        transform: translateY(-80px);
    }

}


/* =========================================================
   HERO
========================================================= */

.hero {
    text-align: center;
    padding: 55px 20px 35px;
    position: relative;
}

.hero-badge {

    display: inline-block;

    padding: 8px 18px;

    border-radius: 50px;

    background: rgba(99,102,241,0.12);

    border: 1px solid rgba(129,140,248,0.35);

    color: #a5b4fc;

    font-size: 13px;

    font-weight: 600;

    letter-spacing: 1px;

    margin-bottom: 20px;

    box-shadow: 0 0 25px rgba(99,102,241,0.12);

}

.hero-title {

    font-size: 54px;

    font-weight: 800;

    margin: 0;

    line-height: 1.1;

    background: linear-gradient(
        90deg,
        #ffffff,
        #a5b4fc,
        #c084fc,
        #ffffff
    );

    background-size: 250% auto;

    -webkit-background-clip: text;

    -webkit-text-fill-color: transparent;

    animation: gradientMove 5s linear infinite;

}

@keyframes gradientMove {

    0% {
        background-position: 0% center;
    }

    100% {
        background-position: 250% center;
    }

}

.hero-subtitle {

    color: #94a3b8;

    font-size: 17px;

    margin-top: 15px;

    max-width: 650px;

    margin-left: auto;

    margin-right: auto;

}


/* =========================================================
   MAIN GLASS CARD
========================================================= */

.main-card {

    position: relative;

    margin: 10px 20px 30px;

    padding: 30px;

    border-radius: 28px;

    background: rgba(15,23,42,0.72);

    backdrop-filter: blur(25px);

    -webkit-backdrop-filter: blur(25px);

    border: 1px solid rgba(148,163,184,0.15);

    box-shadow:

        0 30px 80px rgba(0,0,0,0.45),

        inset 0 1px 0 rgba(255,255,255,0.05);

}


/* =========================================================
   LANGUAGE BAR
========================================================= */

.language-box {

    background: rgba(30,41,59,0.65);

    border: 1px solid rgba(148,163,184,0.12);

    border-radius: 20px;

    padding: 18px;

}


/* =========================================================
   LABELS
========================================================= */

label {

    color: #cbd5e1 !important;

    font-weight: 600 !important;

}


/* =========================================================
   DROPDOWNS
========================================================= */

.dropdown-custom {

    border-radius: 15px !important;

}


/* =========================================================
   TEXT BOX
========================================================= */

textarea {

    background: rgba(15,23,42,0.75) !important;

    color: #f8fafc !important;

    border: 1px solid rgba(148,163,184,0.18) !important;

    border-radius: 18px !important;

    font-size: 16px !important;

    line-height: 1.6 !important;

    padding: 18px !important;

    transition: all 0.3s ease !important;

}

textarea:focus {

    border: 1px solid #818cf8 !important;

    box-shadow:
        0 0 0 3px rgba(99,102,241,0.12),
        0 0 30px rgba(99,102,241,0.08) !important;

}


/* =========================================================
   INPUT / OUTPUT HEADINGS
========================================================= */

.section-title {

    color: #e2e8f0;

    font-size: 16px;

    font-weight: 700;

    margin: 8px 0 10px;

}


/* =========================================================
   TRANSLATE BUTTON
========================================================= */

.translate-button {

    height: 58px !important;

    border: none !important;

    border-radius: 16px !important;

    background: linear-gradient(
        135deg,
        #6366f1,
        #8b5cf6,
        #a855f7
    ) !important;

    color: white !important;

    font-size: 17px !important;

    font-weight: 700 !important;

    box-shadow:
        0 10px 30px rgba(99,102,241,0.25);

    transition: all 0.3s ease !important;

}

.translate-button:hover {

    transform: translateY(-3px) scale(1.01);

    box-shadow:
        0 15px 40px rgba(99,102,241,0.40);

}


/* =========================================================
   SECONDARY BUTTONS
========================================================= */

.secondary-button {

    border-radius: 14px !important;

    background: rgba(51,65,85,0.55) !important;

    border: 1px solid rgba(148,163,184,0.15) !important;

    color: #cbd5e1 !important;

    font-weight: 600 !important;

    transition: all 0.25s ease !important;

}

.secondary-button:hover {

    background: rgba(71,85,105,0.7) !important;

    transform: translateY(-2px);

}


/* =========================================================
   CHARACTER COUNTER
========================================================= */

.counter {

    text-align: right;

    color: #64748b;

    font-size: 12px;

    margin-top: 5px;

}


/* =========================================================
   FEATURES
========================================================= */

.features {

    display: grid;

    grid-template-columns:
        repeat(3, 1fr);

    gap: 15px;

    margin-top: 25px;

}

.feature {

    padding: 18px;

    border-radius: 18px;

    background: rgba(30,41,59,0.45);

    border: 1px solid rgba(148,163,184,0.10);

    text-align: center;

    transition: all 0.3s ease;

}

.feature:hover {

    transform: translateY(-4px);

    background: rgba(51,65,85,0.55);

}

.feature-icon {

    font-size: 25px;

    margin-bottom: 8px;

}

.feature-title {

    color: #e2e8f0;

    font-weight: 700;

    font-size: 14px;

}

.feature-text {

    color: #64748b;

    font-size: 12px;

    margin-top: 4px;

}


/* =========================================================
   FOOTER
========================================================= */

.footer {

    text-align: center;

    padding: 25px 20px 40px;

    color: #64748b;

    font-size: 13px;

}

.footer strong {

    color: #a5b4fc;

}


/* =========================================================
   MOBILE
========================================================= */

@media (max-width: 700px) {

    .hero-title {
        font-size: 38px;
    }

    .main-card {
        margin: 10px;
        padding: 18px;
    }

    .features {
        grid-template-columns: 1fr;
    }

}

"""


# ============================================================
# GRADIO UI
# ============================================================

with gr.Blocks(
    css=custom_css,
    title="AI Language Translator"
) as app:


    # --------------------------------------------------------
    # HERO
    # --------------------------------------------------------

    gr.HTML("""
        <div class="hero">

            <div class="hero-badge">
                ✦ CODEALPHA AI INTERNSHIP PROJECT
            </div>

            <h1 class="hero-title">
                🌍 AI Language Translator
            </h1>

            <p class="hero-subtitle">
                Translate your text instantly across multiple languages
                with a simple and intelligent translation experience.
            </p>

        </div>
    """)


    # --------------------------------------------------------
    # MAIN CARD
    # --------------------------------------------------------

    with gr.Column(elem_classes="main-card"):


        # Language selection

        gr.HTML("""
            <div class="section-title">
                🌐 Choose Languages
            </div>
        """)

        with gr.Row(elem_classes="language-box"):

            source_language = gr.Dropdown(
                choices=list(languages.keys()),
                value="English",
                label="From",
                elem_classes="dropdown-custom"
            )

            swap_button = gr.Button(
                "⇄",
                scale=0,
                min_width=65,
                elem_classes="secondary-button"
            )

            target_language = gr.Dropdown(
                choices=list(languages.keys()),
                value="Hindi",
                label="To",
                elem_classes="dropdown-custom"
            )


        # ----------------------------------------------------
        # INPUT
        # ----------------------------------------------------

        gr.HTML("""
            <div class="section-title">
                ✍️ Enter Your Text
            </div>
        """)

        text_input = gr.Textbox(
            placeholder="Type or paste your text here...",
            lines=7,
            show_label=False
        )

        counter = gr.Markdown(
            "0 characters",
            elem_classes="counter"
        )


        # ----------------------------------------------------
        # ACTION BUTTONS
        # ----------------------------------------------------

        with gr.Row():

            translate_button = gr.Button(
                "✨  Translate Text",
                elem_classes="translate-button",
                scale=4
            )

            clear_button = gr.Button(
                "🗑️ Clear",
                elem_classes="secondary-button",
                scale=1
            )


        # ----------------------------------------------------
        # OUTPUT
        # ----------------------------------------------------

        gr.HTML("""
            <div class="section-title">
                🔤 Translation Result
            </div>
        """)

        output = gr.Textbox(
            placeholder="Your translated text will appear here...",
            lines=7,
            show_label=False
        )


        # ----------------------------------------------------
        # FEATURES
        # ----------------------------------------------------

        gr.HTML("""
            <div class="features">

                <div class="feature">

                    <div class="feature-icon">⚡</div>

                    <div class="feature-title">
                        Fast Translation
                    </div>

                    <div class="feature-text">
                        Get results quickly
                    </div>

                </div>


                <div class="feature">

                    <div class="feature-icon">🌎</div>

                    <div class="feature-title">
                        Multiple Languages
                    </div>

                    <div class="feature-text">
                        Translate across languages
                    </div>

                </div>


                <div class="feature">

                    <div class="feature-icon">✨</div>

                    <div class="feature-title">
                        Simple Interface
                    </div>

                    <div class="feature-text">
                        Easy for everyone
                    </div>

                </div>

            </div>
        """)


    # --------------------------------------------------------
    # FOOTER
    # --------------------------------------------------------

    gr.HTML("""
        <div class="footer">

            Built with Python + Gradio + Translation API

            <br><br>

            <strong>CodeAlpha Artificial Intelligence Internship</strong>

        </div>
    """)


    # ========================================================
    # EVENTS
    # ========================================================

    translate_button.click(
        fn=translate_text,
        inputs=[
            text_input,
            source_language,
            target_language
        ],
        outputs=output
    )


    swap_button.click(
        fn=swap_languages,
        inputs=[
            source_language,
            target_language
        ],
        outputs=[
            source_language,
            target_language
        ]
    )


    clear_button.click(
        fn=clear_all,
        inputs=[],
        outputs=[
            text_input,
            output,
            source_language,
            target_language
        ]
    )


    text_input.change(
        fn=character_count,
        inputs=text_input,
        outputs=counter
    )


# ============================================================
# LAUNCH
# ============================================================

app.launch(share=True)
```
